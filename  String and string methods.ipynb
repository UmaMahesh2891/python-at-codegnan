#####  String Methods 
 - Case Conversion Methods
    - upper(): It coverts entire String into Uppercase
    - lower(): It coverts entire String into lowercase
    - title(): It converts every word first character into uppercase and remaining into lowercase 
    - swapcase(): IT converts lower into upper and upper into lower
    - capitalize():It conerts first character of string into uppercase and remaining into lower

```python
String = "The ocean is blue and sky is BLUE"
lower = String.lower()
print(lower)
print(String.upper())
print(String.title())
print(String.swapcase())
print(String.capitalize())
```

```python
# White Space Remove methods
# strip(): IT removes the white space to both sides of string 
# Lstrip(): It removes leading white space  or left side space  
# rstrip(): It removes lagging white space or right side space
Str = "         Codegnan        "
print(Str)
print(Str.strip())
print(Str.lstrip())
print(Str.rstrip())
```

```python
# White Space Remove methods
# strip(): IT removes the white space to both sides of string 
# Lstrip(): It removes leading white space  or left side space  
# rstrip(): It removes lagging white space or right side space
Str = "@@@@@@@@@@@ Code @  gnan @@@@@@@@@"
print(Str)
print(Str.strip("@"))
print(Str.lstrip("@"))
print(Str.rstrip("@"))
```

```python
# split(): It returns the list of 
String = "The ocean is blue and sky is BLUE 1234@"
print(String.split())
print(String.split('blue'))
print(String.split('is',1))
print(String.split('z'))
print(String.split('@'))
```

```python
# Indexing Finding Methods
# index(sub_string): It returns the first occurance index of substring if string exits otherwise "valueError"
# find(sun_string) : It returns the first occurance index of substring if string exits otherwise "valueError"
String ="The ocean is blue and sky is blue"
print(String.index("o"))
print(String.index("is"))
print(String.index("is",20)) # it start checking from 20th index
print(String.index("z"))
```

```python
# Indexing Finding Methods
# index(sub_string): It returns the first occurance index of substring if string exits otherwise "valueError"
# find(sun_string) : It returns the first occurance index of substring if string exits otherwise "-1"
String ="The ocean is blue and sky is blue"
print(String.index("o"))
print(String.find("is"))
print(String.find("is",20))
print(String.find("z"))
```

```python
# replace(old, new): It replaces the old_string with new_string in string
String ="The ocean is blue and sky is blue"
print(String.replace('blue','Red'))
print(String.replace('blue','Red',1)) # it replaces only 1st occurance
```

```python
# "sep".join(iterable): It joins the all iterable elements with sepeartor
list = ['The', 'ocean', 'is', 'blue', 'and', 'sky', 'is', 'BLUE']
print("".join(list))
print(" ".join(list))
print("_".join(list))
```

### string checkoing Methods
 - islower(): It returns True when entire Str is in lowercase 
 - isupper(): It returns True when entire Str is in uppercase
 - istitle():  It returns True when  Str is in title format 
 - isalpha(): It returns True when  Str contains Alphabets(i.e,lower and uppercase) 
 - isalnum(): it returns True when str either Alphabets or digits
 - isdigit(): it returns True when str contain onlu digits 
 - startswith(): it returns True when String startwith the sub_string
 - endswith(): it returns True when string endswith the sub_string

```python
print("hello".islower())
print("HEllo".isupper())
print("HELLO".isupper())
print("HELLO WORLD".startswith("H"))
print("hello".endswith("o"))
```
