```python
# List Comprehension
# syntax :[exp for iterator in iteable if condition]
# return 1-n numbers as list list Comprehension
n = int(input())
res = [x for x in range(1, n+1)]
print(res)
```

```python
# Find the first n even number
n = int(input())
res = [x for x in range(1, n+1) if x % 2 == 0]
print(res)
```

```python
# Find the first n even number
n = int(input())
res = [2 * x for x in range(n)]
print(res)
```

```python
# Find the first n even number
n = int(input())
res = [x for x in range(0, 2*n, 2)]
print(res)
```

```python
# Find even or odd for list of integers
arr = [20,34,33,55,23,54,77]
res = [ "odd" if x&1 else "even" for x in arr]
print(res)
```

```python
arr = [20,34,33,55,23,54,77]
res = [ "even"if x%2==0 else "odd" for i in arr]
print(res)
```

```python
# even - add 1
# odd - add 0.5
arr = [20,34,33,55,23,54,77]
res = [ x+0.5 if x&1 else x+1 for x in arr]
print(res)
```

```python
# find the one dice combinations
print([x for x in range(1,7)])
```

```python
# find the two dice combinations
print([(i, j) for i in range(1,7) for j in range(1,7)])
```

```python
#  find the two dice combinations if any dice face is 4
print([(i,j) for i in range(1,7) for j in range(1,7) if i ==4 or j==4])
```
