```python
class Mobile:

    company = "iPhone"

    def __init__(self, model, ram, rom, price):
        self.model = model
        self.ram = ram
        self.rom = rom
        self.price = price

    def about_phone(self):
        return f"The {self.model} has {self.ram} GB RAM and {self.rom} GB ROM"

    def display(self):
        return "This is Samsung company display"

    # Class method
    @classmethod
    def update_company(cls, new_name):
        cls.company = new_name
        return "Company updated successfully"

    # Static method
    @staticmethod
    def charger_provide(cable=False):
        return "Adapter + Cable" if cable else "Only Adapter"


# Object creation
iphone16 = Mobile("iPhone 16", 16, 256, 125000)
iphone17 = Mobile("iPhone 17", 8, 256, 120000)

print(iphone16.about_phone())
print(iphone16.update_company("iPhone Pro"))
print(Mobile.company)
print(Mobile.charger_provide())
print(Mobile.charger_provide(True))
iphone17.about_phone()
```

```python
class Account:
    Bank_name = "Mahesh"

    def __init__(self, username, balance, password):
        self.username = username          # public
        self._balance = balance           # protected
        self.__password = password        # private

    # Getter method
    def get_balance(self):
        return self._balance

    # Setter method
    def update_password(self, new_password):
        self.__password = new_password
        return "Password updated successfully"

    # Getter method
    def get_password(self):
        return self.__password

    # Withdraw method
    def withdraw(self, withdraw_amount):
        if self._balance >= withdraw_amount:
            self._balance -= withdraw_amount
            return f"{withdraw_amount} withdrawn successfully"
        return "Insufficient balance"

    # Deposit method
    def deposite(self, deposite_amount):
        self._balance += deposite_amount
        return (
            f"{deposite_amount} deposited successfully.\n"
            f"Current balance: {self._balance}"
        )

    # Private method
    def __total_users(self):  
        return "Total users: 100"


user = Account("Mahesh", 5000, "Uma@123")

print(user.username)
print(user.get_balance())
print(user.deposite(5000))
print(user.update_password("mahi@123"))
print(user.get_password())

# Access protected member
print(user._balance)

# Access private method using name mangling
print(user._Account__total_users())
```

```python
# Inherete the parent class constructor in child class by using super()
# super(): This function is user to get the parent class implementation
class person:
    def __init__(self,name,age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old and gender is {self.gender}"
# child class
class Student(person):
    def __init__(self, stdid, name, age, gender):
        self.stdid = stdid
        super().__init__(name, age, gender)
    def percentage(self, marks:list):
        return sum(marks)/len(marks)
        
    
s = Student(1001,"Ravi", 20, "male")
print(s.name)
print(s.details())
print(s.percentage([90,91,89]))
```
