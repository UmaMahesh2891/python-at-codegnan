```python
percentage = [34,35,54,65,67]
if all(percentage)>35:
    print("pass")
else:
    print("not all pass")
```

```python
percentage = [34,35,54,65,67]
lst = []
for num in percentage:
    if num < 35:
        lst.append(False)
    else:
        lst.append(True)
print(lst)
if all(lst):
    print("All stuidents pass")
else:
    print("Al students Not Passed")
```
