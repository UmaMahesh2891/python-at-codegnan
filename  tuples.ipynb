```python
### Dictionary
# dict can't allow list qs key
# if dict contains multliple items with key, in the case it updates first item key value with last defined item value
details = {'name':'ravi', 'age':20, 'course':'python','marks':[87,99,89], 'age':50}
print(details)
```

```python
# access values by using keys
details = {'name':'ravi', 'age':20, 'course':'python','marks':[87,99,89], 'age':50}
print(details['course'])
print(details['age'])

# Updating the Dicitionary Value
details['age'] += 1
details['course'] = 'java'
print(details)

# item assigment
details['location'] = "India"    # It adds new item to dictionary
print(details)
```

## dictionary methods
 - get value
   - get(key, default_val): It returns the value if key exists in dictonary otherwisedefault value
 - update values and add items to dictionary
  - update(new_dictonary): Dictonary updates with new new_dictionary Items
 - Rempove items from dictionary
   - pop(key): It returns value and remove item if key is exists in dictionary otherwise keyError
   - popitem(): it returns tuple of key value and removes the item from dict
 - get keys, value and items from dict
   - keys():it returns list of dict keys
   - values(): it returns list of dict values
   - items(): it returns list of tuple key-value pairs 

```python
details = {'name':'ravi', 'age':20, 'course':'python','marks':[87,99,89], 'age':50}
print(details['location'])
print(details[0])
```

```python
## 
details = {'name':'ravi', 'age':20, 'course':'python','marks':[87,99,89], 'age':50}
print(details.get('name'))
print(details.get('location')) # deosn't exists 
print(details.get('age', 18))
print(details.get('location', "Location key not exists")) # here we are default value

      
```

```python
details = {'name':'ravi', 'age':20, 'course':'python'}
d =  {'marks':[87,99,89], 'age':50}
details.update(d)
print(details)
```
