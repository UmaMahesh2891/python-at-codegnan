#### List Methods
- Elements Adding Methods
 - append(item): It adds item at enf of the list
 - extand(iterable): It adds all iterable elements at end of the list 
 - insert(pop, item):  This method adds the item at specified position


```python
lst = [1,2,3,4]
lst.append(7)
lst.append('code')
print(lst)
lst.append([10,20,30])
print(lst)
lst.extend([100,200,300])
print(lst)
lst.insert(1,2000)
print(lst)
```

- Elements remove methos
 - remove(val): It removes first occurance of value if value exists in list otherwise it raises 'ValueError'
 - pop(): It Removes and returns the last of element of list
 -pop(pos): It removes and returns the specified position element from list
 -clear(): it removes all element from list
 -del : it python keyword,used to delete values


```python
lst = [10,20,30,40,20]
lst.remove(20)
print(lst)
val = lst.pop()
print(val, lst)
#lst.remove(100)  list.remove(x): x not in list
val2 = lst.pop(1)
print(val2, lst)
#print(lst.pop(10)) pop index out of range
del lst
print(lst)
lst.clear()
print(lst)

```

```python
# index(val) : It teruns the index of frist occurance of value if value exists in list otherwise value error
# count(val) : It returns the total frequency of value\
lst = [11,2,12,4,14,2,2,2]
ind = lst.index(2)
print(ind)
print(lst.count(2))
print(lst.count(100))
print(lst.index(100))
```
