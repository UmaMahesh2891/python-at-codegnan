```python
# find the factors of a numbers
n = int(input())
factors = []

for i in range (1, n+1):
    if n % i == 0:
        factors.append(i)
print(factors)
```

```python
n = int(input())
factors = []

for i in range (1, n+1):
    if n % i == 0:
        factors.append(i)
print(factors)
if len(factors) == 2:
    print(f"{n} is prime")
else:
    print(f"{n} is not prime")
```

```python
num = int(input())
for i in range(2, num//2+1):
    if num % i == 0:
        print(f"{num} is not prime")
        break
else:
    print(f"{num} is prime")
    
```

```python
# find the reverse of a string
string = input()
res = ""
for char in string:
    res = char + res
    print(res)
```

```python
string = input()
res = ""
for char in string:
    res = char + res
print(res)
if res == string:
    print(f"{string} is palindrome")
else:
    print(f"{string} is not palindrome")
```
