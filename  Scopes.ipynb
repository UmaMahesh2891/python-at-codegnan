1. Local Scope

-A variable created inside a function belongs to the local scope.

 -Exists only while the function is executing.
 -Cannot be accessed outside the function. 



```python
# Local Scope:
# We can access the local scope variable within the spoce,we can't access in a global scope

def fun(x):
    x = x+1 # x is local scope
    return x
a = 10
res = fun(a)
print(x)
```

2. Enclosing Scope (Nonlocal Scope)

  - This scope appears in nested functions.

  - Variables belong to the outer function but are used by the inner function



```python
# Enclosing Scope
def outer(x):
    y = 10 # Enclose Scope
    def inner():
        a = 10 # Local scope
        z = y+a
        print(f"this is local scope:{z}")
    inner()
    y = y+x
    print(f"this is enclosing scope:{y}")
n = 5
outer(n)
```

```python
# Enclosing Scope
def outer(x):
    x = 20 # Enclose Scope
    def inner():
        x = 20 # Local scope
        z = x+x
        print(f"this is local scope:{z}")
    inner()
    y = x+x
    print(f"this is enclosing scope:{y}")
n = 5
outer(n)
```

```python
# Global Scope
x = 10
def outer():
    y = 20
    def inner():
        z = 30
        res = z+y+x
        print(f"This is local scope: {res}")

    res = y+x # 30
    inner()
    print(f"This is Enclosing Scope:{res}")
outer()
```

```python
# Global Scope
x = 10
def outer():
    y = 20
    def inner():
        z = 30
        res = z+y+x
        print(f"This is local scope: {res}")

    y = y+x # 30
    inner()
    print(f"This is Enclosing Scope:{y}")
outer()
```

```python
# Trying to update the global variable inside the local scope
x = 10
def outer():
    x = 20
    x = x+1
    print("This is a local scope val:",x)
outer()
print("this is a global val:", x)
```

```python
# update the global scope variable inside the local scope by using global keyword
x = 10
def outer():
    global x
    x = x+1
    print("This is a local scope val:",x)
outer()
print("this is a global val:", x)
```

```python
# Trying to update enclosing scope variable in local scope
x = 10
def outer():
    y = 20 # Enclosing
    def inner():
        nonlocal y
        z = 30 # local scope
        y = y+1
        print(f"This is local scope: {y}")

    y = y+x # 30
    inner()
    
    print(f"This is Enclosing Scope:{y}")
outer()
```
