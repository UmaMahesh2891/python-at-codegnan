```python
name = 'Ravi'
age = 20

print("My name " + name + " and i am a " + str(age) + " years old")
```

```python
## Output Formatting
## 1. Old Style Formatting (usign comma)
name = 'Ravi'
age = 20

print("My name", name, "and i am a", str(age), "years old")
```

```python
## 1. Access Specifiers
## integer - %d
## float - %f
## string - %s
name = 'Ravi'
age = 20

print("My name %s and i am a %d years old"%(name, age))
print("My name %s and i am a %d years old"%(age, name))
```

```python
## dot formatting
name = 'Ravi'
age = 20

print("My name {} and i am a {} years old".format(name, age))
print("My name {} and i am a {} years old".format(age, name))
```

```python
## 4 . f-format
name = 'Ravi'
age = 20
percentage = 98.2568

print(f"My name {name}. and i am a {age} years old and my percentage is {percentage}")
print(f"My name {name}. and i am a {age} years old and my percentage is {percentage:.2f}")
```
