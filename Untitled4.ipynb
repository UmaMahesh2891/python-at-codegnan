```python
'''
D. Difference
time limit per test1 second
memory limit per test256 megabytes
Given four numbers A, B, C and D. Print the result of the following equation :

 X = (A * B) - (C * D).

Input
Only one line containing 4 separated numbers A, B, C and D ( - 105  ≤  A, B, C, D  ≤  105).

Output
Print "Difference  =  " without quotes followed by the equation result.

Examples
InputCopy
1 2 3 4
OutputCopy
Difference = -10
InputCopy
2 3 4 5
OutputCopy
Difference = -14
InputCopy
4 5 2 3
OutputCopy
Difference = 14
'''
a, b, c, d = map(int, input().split())
x = (a * b) - (c * d)
print(f"Difference = {x}")

```

```python
n = 7
k = 0
for i in range(n):
    for j in range(n):
        if k < 26:
            ch = chr(65 + k)      # uppercase A-Z
        else:
            ch = chr(97 + (k - 26))  # lowercase a-z
        print(ch, end=" ")
        k += 1
    print()
```
