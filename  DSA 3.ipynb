```python
## print the natural numbers using while loop
i = 1
while i<= 10:
    print(i) 
    i += 1
    
```

```python
## sum of numbers in list
sum = 0
l = [10,2,25,26,7]
for ele in l:
    sum += ele
    
print(sum)
```

```python
## find the min element from list 
mini = float('inf')
l = [7,5,19,12,30,3]
for num in l:
    if num < mini:
        mini = num
print(mini)
        
```

```python
## find the max element from list
mini = float('-inf')
l = [7, 5, 19, 12, 30, 3]
for num in l:
    if num > mini:      # ✅ compare to the mini variable
        mini = num
print(mini)
```

```python
even, odd = 2, 1
for r in range(1,4):
    for c in range(1,4):
        if r % 2 == 0:
            print(odd, end="\t")
            odd += 2
        else:
            print(even, end="\t")
            even += 2
    print()
```
