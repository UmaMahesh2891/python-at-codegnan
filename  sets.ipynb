### set Methods
 - Adding elements to set
   - add(ele): it adds elements to set if element not exists in set,otherwise nothing
   - update(iterable): it adds all iterable elements to set
 - Remove elements from set
   - remove(ele):it removes the element from the set if element exixts in set otherwise 'KeyError'
   - discard(ele): it removes the element from set if element exists in set otherwise nothing
   - pop():It removes and returns the elements from set
 - set operations
    - union(anthor_sets): It returns the all common elements from all sets
    - intersection(anthoe_sets):it returns the common elements from the all sets
    - difference(anthor_sets):it returns the sets elements those are not lies in anthor_sets
    - symmetric difference(anthor_sets) : It returns the Non common elements fron all sets 

```python
s = {1,2,3,4,5,6,7}
s.add(99)
s.add(2.1)
s.add(2)

print(s)
s.remove(1)
print(s)
s.discard(2.1)
print(s)
#s.remove(1)
print(s)
s.discard(1)
print(s)
```

```python
s1 = {1,2,3,4,88,99,77}
s2 = {77,99,88,4,7,8,9}
s3 = {77,10,6,5}
print(s1.union(s2))
print(s1.union(s2, s3))
print(s1.intersection(s2))
print(s1.intersection(s2,s3))
print(s1.difference(s2))
print(s1.defference(s2,s3))
```

```python
s1 = {1,2,3,4,88,99,77}
s2 = {77,99,88,4,7,8,9}
s3 = {77,10,6,5}
print(s1 | s2) #  | - union
print(s1 | s2 | s3)
print(s1 & s2) # & - intersection
print(s1 & s2 &s3)
print(s1 - s2) # - fro difference
print(s1 - s2 - s3)
print(s1 ^ s2)
```

```python
s1 = {1,2,3}
s2 = {2,3,4,5}
s1.intersection_update(s2)
print(s1)
s1.difference_update(s2)
print(s1, s2)
s1.symmertic_difference_update(s2)
print(s1)
```
