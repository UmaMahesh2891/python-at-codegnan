```python
## Find the weather the number is prime number 
def prime_check(n:int)->bool:
    if n < 2:
        return False
    for i in range(2, n//2+1):
        if n % i == 0:
            return False
    return True
num = int(input())
print(prime_check(num))
    
```

```python
# Prime Numbers between 1-100
def prime_check(n:int)->bool:
    if n < 2:
        return False
    for i in range(2, n//2+1):
        if n % i == 0:
            return False
    return True
for j in range(1, 100):
    if prime_check(j):
        print(j, end = " ")
```
